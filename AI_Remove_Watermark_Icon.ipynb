```python
# Mount thư mục kết quả vào Google Drive cá nhân
print('Đang kết nối Google Drive...')
from google.colab import drive
import os, shutil
drive.mount('/content/drive')

# Đảm bảo thư mục cha tồn tại
os.makedirs('/content/app/content/app', exist_ok=True)

target_link = '/content/app/content/app/image_results'
if os.path.exists(target_link) or os.path.islink(target_link):
    if os.path.islink(target_link) or os.path.isfile(target_link):
        os.unlink(target_link)
    else:
        shutil.rmtree(target_link)

os.symlink('/content/drive/MyDrive', target_link)
print('Đã kết nối thư mục kết quả vào Google Drive thành công!')
```
