```python
#mount thư mục kết quả vào google drive cá nhân
#bước này tuy không ép buộc nhưng khá cần thiết
print('Đang kết nối google drive')
from google.colab import drive
import os, shutil
drive.mount('/content/drive')

drive_folder = '/content/drive/MyDrive/AI_Remove_Watermark_Icon'
os.makedirs(drive_folder, exist_ok=True)

app_dir = '/content/app'
os.makedirs(app_dir, exist_ok=True)

local_results = os.path.join(app_dir, 'image_results')
if os.path.islink(local_results) or os.path.isfile(local_results):
    os.unlink(local_results)
elif os.path.isdir(local_results):
    shutil.rmtree(local_results)

os.symlink(drive_folder, local_results)
print('Đã kết nối thư mục kết quả vào Google Drive thành công!')
```
